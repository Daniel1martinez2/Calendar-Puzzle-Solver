# Daily Calendar Puzzle — Demo

This notebook demonstrates the ILP-based solver for the Daily Calendar Puzzle.
Each day, 10 polyomino pieces must tile the board leaving exactly three cells
exposed: the current **month**, **day**, and **weekday** (Spanish labels).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

from calendar_puzzle import Board, Solver, Renderer

board = Board()
solver = Solver(board)
renderer = Renderer(board)
```

## The Board

An 8×7 grid with 50 playable cells — 12 months, 31 days, and 7 weekdays.

```python
renderer.show_board({})
```

## The Pieces

10 polyomino pieces (4–5 cells each, 47 cells total) with all unique orientations.

```python
renderer.show_pieces()
```

## Solve: FEB 24 MAR (martes)

Find a tiling that leaves February, 24, and Tuesday exposed.

```python
result = solver.solve("FEB", 24, "MAR")
targets = board.target_cells("FEB", 24, "MAR")
renderer.show_board(result, targets, show_labels=False)
```

## Solve: AGO 15 VIE (viernes)

```python
result2 = solver.solve("AGO", 15, "VIE")
targets2 = board.target_cells("AGO", 15, "VIE")
renderer.show_board(result2, targets2, show_labels=False)
```

---

The ILP formulation (PuLP + CBC) solves each date in under a second.
